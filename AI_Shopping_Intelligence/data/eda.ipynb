```python
import pandas as pd
import numpy as np  
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from flaml import AutoML
from sklearn.metrics import accuracy_score
```

```python
PATH=r"C:\Users\LENOVO\Desktop\Shopping\AI_Shopping_Intelligence\data\raw\Main_review.csv"
```

```python
df=pd.read_csv(PATH)
df.head()
```

```python
# df.rename(columns={"text_":"text"},inplace=True)
# df["label"]=df["label"].apply(lambda x: '0' if x=="CG" else '1')
# df["label"]=df["label"].astype(int)
```

```python
df.rename(columns={"text_":"text"},inplace=True)
df["label"] = (df["label"] == "CG").astype(int)

```

```python
df.head()
```

```python
df['label'].value_counts()
```

```python
df[df['label']==0]['label']
df.drop(columns=["category","rating"])
```

```python
features=df["text"]
target=df['label']
```

```python
tf_idf=TfidfVectorizer()
x_vectorized=tf_idf.fit_transform(features)
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x_vectorized,
    target,
    test_size=0.2,
    random_state=42
)
# x_test=tf_idf.transform(x_test)
model=AutoML()
```

```python
model.fit(
    x_train,
    y_train,
    task="classification",
    time_budget=300
)
```

```python
model.model
```

```python
model.best_estimator
```

```python
model.best_loss
```

```python
prediction=model.predict(x_test)
accuracy = accuracy_score(y_test, prediction)
print(f"Accuracy: {accuracy:.4f}")
```
